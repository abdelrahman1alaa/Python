# Part 1: If-Conditions & Functions
**1. Check Even or Odd Number:**
*Problem:* Write a function that accepts an integer from the user and prints whether the number is even or odd.

Solution:

```python
def check_even_odd():
    num = int(input("Enter a number: "))
    if num % 2 == 0:
        print("The number is Even")
    else:
        print("The number is Odd")

check_even_odd()
```

# 2. Calculate University Grade
Problem: Write a program that takes a student's score (0 to 100) and prints their grade:

90 or above: Excellent

80 to 89: Very Good

70 to 79: Good

50 to 69: Pass

Below 50: Fail

Solution:

```python
def get_grade():
    score = float(input("Enter grade (0-100): "))
    if score >= 90:
        print("Excellent")
    elif score >= 80:
        print("Very Good")
    elif score >= 70:
        print("Good")
    elif score >= 50:
        print("Pass")
    else:
        print("Fail")

get_grade()
```

# 3. Find Largest Among Three Numbers
Problem: Create a function that accepts 3 numbers from the user and prints the largest one.

Solution:

```python
def find_largest():
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    num3 = float(input("Enter third number: "))
    
    if num1 >= num2 and num1 >= num3:
        print(f"The largest number is: {num1}")
    elif num2 >= num1 and num2 >= num3:
        print(f"The largest number is: {num2}")
    else:
        print(f"The largest number is: {num3}")

find_largest()
```

# 4. Shipping Cost Calculator
Problem: Write a function that calculates shipping cost based on package weight:

Under 2 kg: $20

2 to 10 kg: $50

Over 10 kg: $100

Solution:

```python
def shipping_cost():
    weight = float(input("Enter package weight in kg: "))
    if weight < 2:
        cost = 20
    elif weight <= 10:
        cost = 50
    else:
        cost = 100
    print(f"Shipping cost is: ${cost}")

shipping_cost()
```

**5. Multiplication Table**
Problem: Ask the user for a number, then print its multiplication table from 1 to 10 using a for loop.

Solution:

```python
def multiplication_table():
    num = int(input("Enter a number to show its multiplication table: "))
    for i in range(1, 13):
        print(f"{num} x {i} = {num * i}")

multiplication_table()
```

# 6. Sum of Numbers Up to N
Problem: Read a positive integer N from the user, calculate the sum of all numbers from 1 to N, and print the result.

Solution:

```python
def sum_up_to_n():
    n = int(input("Enter a positive number N: "))
    total = 0
    for i in range(1, n + 1):
        total += i
    print(f"The sum of numbers from 1 to {n} is: {total}")

sum_up_to_n()
```

# 7. Countdown Timer
Problem: Ask the user for a starting number and print the numbers counting down to 0 using a while loop.

Solution:

```python
def countdown():
    start = int(input("Enter starting number for countdown: "))
    while start >= 0:
        print(start)
        start -= 1

countdown()
```

# 9. Count Digits of an Integer
Problem: Ask the user to input an integer, then calculate and print the number of digits in that integer.

Solution:

```python
def count_digits():
    number = int(input("Enter an integer: "))
    count = 0
    temp = abs(number) # Handle negative numbers
    
    if temp == 0:
        count = 1
    else:
        while temp > 0:
            temp //= 10
            count += 1
            
    print(f"Number of digits: {count}")

count_digits()
```

# Part 3: Lists & Tuples
**10. Find Min and Max in a List :**
Problem: Ask the user to input 5 space-separated numbers, store them in a list, and find the maximum and minimum numbers manually without using max() or min().

Solution:

```python
def find_min_max():
    raw_input = input("Enter 5 space-separated numbers: ")
    numbers = [float(x) for x in raw_input.split()]
    
    largest = numbers[0]
    smallest = numbers[0]
    
    for num in numbers:
        if num > largest:
            largest = num
        if num < smallest:
            smallest = num
            
    print(f"Largest number: {largest}")
    print(f"Smallest number: {smallest}")

find_min_max()
```

# 11. Filter Even Numbers
Problem: Ask the user for a list of numbers, filter out odd numbers, and print a new list containing only even numbers.

Solution:

```python
def filter_even_numbers():
    raw_input = input("Enter numbers separated by spaces: ")
    numbers = [int(x) for x in raw_input.split()]
    even_list = []
    
    for num in numbers:
        if num % 2 == 0:
            even_list.append(num)
            
    print(f"Even numbers only: {even_list}")

filter_even_numbers()
```

# 12. Search Item in a Tuple
Problem: Create a tuple containing names of fruits. Ask the user to input a fruit name and check whether it exists in the tuple.

Solution:

```python
def search_in_tuple():
    fruits = ("apple", "banana", "orange", "mango", "grape")
    user_fruit = input("Enter a fruit name to search: ").strip().lower()
    
    if user_fruit in fruits:
        print(f"Yes, {user_fruit} is available!")
    else:
        print(f"Sorry, {user_fruit} is not available.")

search_in_tuple()
```

# 13. Reverse List of Words
Problem: Read space-separated words from the user, store them in a list, and print the words in reverse order.

Solution:

```python
def reverse_words():
    raw_input = input("Enter words separated by spaces: ")
    words = raw_input.split()
    reversed_words = []
    
    for i in range(len(words) - 1, -1, -1):
        reversed_words.append(words[i])
        
    print("Reversed order:", " ".join(reversed_words))

reverse_words()
```
