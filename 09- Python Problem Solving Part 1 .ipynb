# 1️⃣ (Input + If) - Check Number Type

**Question: Write a program that takes a number from the user and determines whether it's positive, negative, or zero.**

```python
num = int(input("Enter a number: "))

if num > 0:
    print("The number is positive")
elif num < 0:
    print("The number is negative")
else:
    print("The number is zero")
```

# 5️⃣ (For Loop) - Sum of Numbers from 1 to N

**Question: Take a number N from the user and calculate the sum of numbers from 1 to N using a For Loop.**

```python
n = int(input("Enter the number N: "))
total = 0

for i in range(1, n + 1):
    total += i

print("The sum is:", total)
```

# 6️⃣ (While Loop) - Countdown

**Question: Write a program that prints a countdown from 10 to 1 using a While Loop.**

```python
count = 10

while count >= 1:
    print(count)
    count -= 1
```

# 7️⃣ (While Loop + If) - Guess the Number

**Question: The program has a secret number = 7. Let the user keep guessing until they get it right.**

```python
secret = 7
guess = int(input("Guess the number: "))

while guess != secret:
    print("Wrong, try again")
    guess = int(input("Guess the number: "))

print("Correct! The number is", secret)
```

# 1️⃣1️⃣ (Lists + If) - Find the Largest Element

**Question: Find the largest number in a list without using the max() function.**

```python
numbers = [15, 42, 8, 99, 23]
biggest = numbers[0]

for num in numbers:
    if num > biggest:
        biggest = num

print("The biggest number is:", biggest)
```

# 1️⃣9️⃣ (Functions) - Function to Calculate Rectangle Area

**Question: Write a function that takes length and width and returns the area of a rectangle, taking the values from the user.**

```python
def rectangle_area(length, width):
    return length * width

l = float(input("Enter the length: "))
w = float(input("Enter the width: "))

area = rectangle_area(l, w)
print("The rectangle area is:", area)
```

# 2️⃣0️⃣ (Functions + If + Lists) - Function to Filter Even Numbers

**Question: Write a function that takes a list of numbers and returns a new list containing only the even numbers.**

```python
def get_even_numbers(numbers_list):
    even_numbers = []
    for num in numbers_list:
        if num % 2 == 0:
            even_numbers.append(num)
    return even_numbers

numbers = [3, 8, 11, 20, 15, 6]
result = get_even_numbers(numbers)

print("The even numbers are:", result)
```
